### 데이터 개수 맞추고 등수 컬럼 "result" 추가

```python
### 컬럼 이름을 사용하여 포함 여부 판별하는 함수 생성 ###
def is_value_in_column(row, a_col, b_col, c_col, d_col):
    a_val = str(row[a_col])  # 숫자 데이터를 문자열로 변환
    b_val = str(row[b_col])
    c_val = str(row[c_col])
    d_val = str(row[d_col])

    if any(val in b_val for val in a_val.split()):
        return 1
    elif any(val in c_val for val in a_val.split()):
        return 2
    elif any(val in d_val for val in a_val.split()):
        return 3
    else:
        return 0


organs['result'] = organs.apply(is_value_in_column, args=('PLAYER_NM', 'N1R_PLAYER_NM', 'N2R_PLAYER_NM', 'N3R_PLAYER_NM'), axis=1)

print(organs)
```
